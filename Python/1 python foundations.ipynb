```python
# Operators in python
# Unary Operators
#Unary operators work with one operand (a single value or variable). They perform operations on just one thing.
x: int = 5
y: int = -x
print(f"y {y}")

# binary not
x: bool = True
y: bool = not x
print(x)
print(y)

# Bitwise not
x: int = 5
y: int = ~x
print(y)

# Arithmetic Operators (+, -, *, /, etc.):
a: int = 30
b: int = 10

print("a + b ", a+b)
print("a - b ", a-b)
print("a * b ", a*b)
print("a / b ", a/b)
# Comparison Operators (==, !=, >, <, etc.):
if 30 <= a <= 50:
  print("The comparison is correct!")
# Logical Operators (and, or):
x: bool = True
y: bool = False

print("x and y = ", x and y)  # False
print("x or y  = ", x or y)   # True
print("not x   = ", not x)    # False
# Assignment Operators (=, +=, -=, etc.):

x= 10
#Additional Assignment
x+=5
print(x)
# Divisional Assignment
x/=5
print(x)

# Modulus
x //= 5
print(x)

# Identity Operator      # is, is not
a: list = [1, 2, 3, 4]
b: list = [1, 2, 3, 4]
c: list = a
print("a is c ", a is c)
print("a is b ", a is b)
print("a is not c ", a is not c)

# Membership operator
my_list: list = [23, 10, 15, 20, 4]
print("my_list ", my_list)
print("5 in my_list ", my_list)
print("40 is not in my_list ", my_list)

print("The list of \
keywords is : ")
```

```python
# Match case statements
#The match-case statement works by comparing an expression against multiple patterns.
def check_status(code):
  match code:
    case 200:
      print("Ok")
    case 200:
      print("Bad Request!")
    case 200:
      print("Not Found")
    case _:
      print("Unknown Status")
check_status(200)
check_status(400)
check_status(404)
```

```python
# *args
def addition(*args):
  s = 0
  for i in args:
    s = s+i
  return s
print(addition(20,30,50,66,59))
```
